```python
import pandas as pd
import os
import numpy as np
import glidertools as gt

import matplotlib.pyplot as plt
import matplotlib.dates as mdates

from datetime import datetime
import cmocean
import netCDF4
from netCDF4 import Dataset
import xarray as xa

import operator as op
```

```python
M2_winds = pd.read_csv('/Users/anjalishah/Desktop/Senior_Thesis/Data/M2_2018_NARR.csv')
M2_winds.index = pd.to_datetime(M2_winds['time'])
M5_winds = M2_winds.drop(columns = ['time'])

fig, ax = plt.subplots(figsize=(30,5))
x = np.array(M5_winds.index)
u= pd.to_numeric(np.array(M5_winds['uwnd'])) #E/W wind, positive is east
v = pd.to_numeric(np.array(M5_winds['vwnd'])) #N/S wind, positive is north

wind_plot = ax.quiver(x, np.zeros(len(x)), u, v, scale=10, scale_units='inches')
ax.set_title('M2 winds 2018')
ax.quiverkey(wind_plot, 0.07, 0.85, U = 10, label ='10 cm/s', labelpos='E') 
ax.set_yticks([])
#ax.set_ylim(-100, 100)
plt.show()
```
